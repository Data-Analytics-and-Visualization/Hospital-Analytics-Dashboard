```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/Hospital Patient .csv")

print("Original shape:", df.shape)

# 1. Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# 2. Remove duplicate records
df = df.drop_duplicates()

# 3. Remove extra spaces from text columns
text_columns = df.select_dtypes(include="object").columns
for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

# 4. Convert dates to proper datetime format
date_columns = ["Admission_Date", "Discharge_Date"]
for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")

# 5. Convert numerical columns to numeric
numeric_columns = ["Age", "Year_of_Admission", "Length_of_Stay", "Total_Cost"]
for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# 6. Check negative values in key numeric columns
negative_check_cols = ["Age", "Length_of_Stay", "Total_Cost"]
for col in negative_check_cols:
    if col in df.columns:
        df.loc[df[col] < 0, col] = None

# 7. Fill missing numerical values with median
for col in numeric_columns:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())

# 8. Fill missing categorical values with mode
for col in text_columns:
    if col in df.columns and df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# 9. Check invalid date relationships (Admission must be before or on Discharge)
if "Admission_Date" in df.columns and "Discharge_Date" in df.columns:
    df = df[df["Discharge_Date"] >= df["Admission_Date"]]

# 10. Reset index
df = df.reset_index(drop=True)

# 11. Final check
print("\nCleaned shape:", df.shape)
print("\nMissing values after cleaning:")
print(df.isnull().sum())

# 12. Save cleaned dataset
df.to_csv("cleaned_hospital_patient_data.csv", index=False)

print("\nData cleaning completed successfully!")
```
